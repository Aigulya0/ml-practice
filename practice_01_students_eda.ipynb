```python
import pandas as pd
import os

url = "https://raw.githubusercontent.com/rashida048/Datasets/master/StudentsPerformance.csv"

os.makedirs("data", exist_ok=True)

df = pd.read_csv(url)

df.to_csv("data/StudentsPerformance.csv", index=False)

print("Датасет успешно загружен!")
print("Количество строк:", df.shape[0])
print("Количество столбцов:", df.shape[1])

display(df.head())
```

```python
print("Размер таблицы:", df.shape)
print("Строк:", df.shape[0])
print("Столбцов:", df.shape[1])
```

```python
print("Типы данных:")
print(df.dtypes)
```

```python
print("Количество пропусков:")
print(df.isnull().sum())
```

```python
print("Базовая статистика:")
display(df.describe())
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(9, 5))

sns.histplot(
    data=df,
    x="math score",
    bins=15,
    kde=True,
    color="skyblue"
)

plt.title("Distribution of Math Scores")
plt.xlabel("Math Score")
plt.ylabel("Number of Students")

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.countplot(
    data=df,
    x="test preparation course",
    color="lightgreen"
)

plt.title("Test Preparation Course")
plt.xlabel("Preparation Course")
plt.ylabel("Number of Students")

plt.tight_layout()
plt.show()
```

```python
print("Количество студентов по подготовке:")
display(df["test preparation course"].value_counts())

print("Средние оценки по группам:")
display(
    df.groupby("test preparation course")[
        ["math score", "reading score", "writing score"]
    ].mean().round(2)
)

print("Корреляция чтения и письма:")
print(
    df["reading score"].corr(df["writing score"])
)
```

## Выводы по анализу данных

**Наблюдение 1. Распределение оценок**

Большинство студентов получили оценки по математике в среднем и высоком диапазоне. Очень низкие оценки встречаются реже.

**Наблюдение 2. Подготовка к экзаменам**

Количество студентов, не проходивших подготовительный курс, больше количества студентов, которые его прошли. При этом у студентов, прошедших подготовку, средние результаты по некоторым предметам выше.

**Наблюдение 3. Связь между предметами**

Оценки по чтению и письму имеют сильную положительную корреляцию. Это означает, что студенты с высокими оценками по чтению обычно также получают высокие оценки по письму.
